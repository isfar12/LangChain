```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_tavily import TavilySearch
from langchain.agents import tool

@tool
def tavily_search(query: str) -> str:
    """Run a search query on Tavily and return the results."""
    search = TavilySearch(max_results=1)
    results = search.invoke(query)
    p=[ result["content"] for result in results["results"] ]
    " ".join(p)
    return p

tools=[tavily_search]
```

```python
answer = tavily_search.invoke("Football matches today")
```

```python
from langchain_groq import ChatGroq

model=ChatGroq(
    model="openai/gpt-oss-20b"
)
```

```python
from langchain import hub

prompt=hub.pull("hwchase17/openai-functions-agent")
prompt
```

```python
from langchain.agents import create_tool_calling_agent
from langchain.agents import AgentExecutor

tool_agent=create_tool_calling_agent(model,tools,prompt)

agent_executor = AgentExecutor(agent=tool_agent,tools=tools,verbose=True)

result=agent_executor.invoke({"input": "Football matches today"})
result["output"]
```
