```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from qdrant_client import QdrantClient, models
QDRANT_URL = os.getenv("QDRANT_URL")
QDRANT_API_KEY = os.getenv("QDRANT_API_KEY")
client = QdrantClient(url=QDRANT_URL, api_key=QDRANT_API_KEY)
```

```python
collection_name = "documents"
model_name = "BAAI/bge-small-en-v1.5"
client.recreate_collection(
    collection_name=collection_name,
    vectors_config=models.VectorParams(size=768, distance=models.Distance.COSINE)
)
```

```python
import uuid
from langchain.schema import Document

documents = [
    Document(id=str(uuid.uuid4()), page_content="Qdrant is a vector search engine that helps you store and search vectors efficiently."),
    Document(id=str(uuid.uuid4()), page_content="Retrieval-augmented generation (RAG) is a method where the language model retrieves relevant information from a database to help answer queries."),
    Document(id=str(uuid.uuid4()), page_content="FastAPI is a modern web framework for building APIs with Python 3.7+ based on standard Python type hints."),
    Document(id=str(uuid.uuid4()), page_content="Machine learning is a subset of artificial intelligence that uses data to train algorithms and improve performance over time.")
]
```

```python
from langchain_qdrant import Qdrant
from langchain_huggingface import HuggingFaceEmbeddings

embedding_function = HuggingFaceEmbeddings(
    model_name="intfloat/e5-base-v2",
    )
```

```python
qdrant_store = Qdrant.from_existing_collection(
    embedding_function,
    collection_name="documents",
    url=QDRANT_URL
)

print("Qdrant vector store initialized successfully!")
```

```python
qdrant_store.add_documents(documents)
```

```python
query = "What is RAG"
results = qdrant_store.similarity_search(query, k=1)
for result in results:
    print(result.page_content)
```
