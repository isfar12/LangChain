```python
from langchain.text_splitter import NLTKTextSplitter, SpacyTextSplitter

sample_text = """
LangChain is a framework for building applications powered by language models.
It provides tools for prompt management, chaining, memory, and retrieval.
Dr. Smith went to Washington, D.C. to attend a conference.
Meanwhile, the company released v2.1.0 of their software in Jan. 2024.
"""

# --- NLTK Splitter ---
nltk_splitter = NLTKTextSplitter(chunk_size=80, chunk_overlap=20)  # chunk_size in characters
nltk_chunks = nltk_splitter.split_text(sample_text)

print("=== NLTKTextSplitter ===")
for i, chunk in enumerate(nltk_chunks, 1):
    print(f"{i}. {chunk}")
print("\n")
```

```python
# --- spaCy Splitter ---
spacy_splitter = SpacyTextSplitter(pipeline="en_core_web_sm", chunk_size=80,chunk_overlap=20)
spacy_chunks = spacy_splitter.split_text(sample_text)

print("=== SpacyTextSplitter ===")
for i, chunk in enumerate(spacy_chunks, 1):
    print(f"{i}. {chunk}")
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=80,
    chunk_overlap=20
)
recursive_chunks = recursive_splitter.split_text(sample_text)

print("=== RecursiveCharacterTextSplitter ===")
for i, chunk in enumerate(recursive_chunks, 1):
    print(f"{i}. {chunk}")
```
